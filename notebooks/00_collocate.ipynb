```python
import pathlib
import site

import pandas as pd
import xarray as xr
from dask.diagnostics.progress import ProgressBar

from highres_ta import colloc

site.addsitedir("../scripts")
import train_quantile_ensemble as tqe
```

```python
config = tqe.load_config(pathlib.Path("../scripts/quantile_ensemble_config.yaml"))
```

```python
data_sources = colloc.read_data_sources("../data/data_sources.yaml")

fname = "../data/raw/GLODAPv2023/GLODAPv2.2023_surface.pq"
glodap_surface_unmatched_all_years = pd.read_parquet(fname).assign(
    lon=lambda x: (x.lon - 180) % 360 - 180  # fix astray lons
)
```

```python
df = tqe.prepare_data(config)
```

```python
years = glodap_surface_unmatched_all_years.time.dt.year
unique_years = years.value_counts().sort_index().index.values
```

```python
for year in unique_years:
    sname = f"../data/matched/GLODAPv2023/GLODAPv2.2023_surface_matched_{year}.pq"
    
    if year < 1982:
        print("before remote sensing era, skipping...")
        continue
    
    if pathlib.Path(sname).exists():
        print(f"File {sname} already exists, skipping...")
        continue
    else:
        print(f"Collocating GLODAP surface data for year {year}...")

    mask = years == year
    glodap_current_year = glodap_surface_unmatched_all_years[mask]
    glodap_surface_matched_current_year = colloc.collocate(glodap_current_year, data_sources)
    
    glodap_surface_matched_current_year.to_parquet(sname)
```
