```python
import matplotlib.pyplot as plt
import numpy as np
from matplotlib.patches import Patch
# Reset to Matplotlib defaults


# Dataset information:
# (Name, Start Year, End Year, Category)

datasets = [
    ("GLODAP", 1972, 2021, "Training"),
    ("ESA SSS CCI", 2010, 2022, "Inference"),
    ("SODA", 1980, 2023, "Inference"),
    ("CMEMS MultiObs", 1993, 2025, "Sensitivity experiments"),
    ("CMEMS GLORYS", 1993, 2025, "Sensitivity experiments"),
    ("ESA SST CCI", 1980, 2024, "Inference"),
    ("OISST", 1981, 2024, "Sensitivity experiments"),
    ("CMEMS SSH", 1993, 2024, "Training"),
    ("CMEMS SSH", 1993, 2024, "Inference"),
    ("GLOBCOLOUR", 1997, 2025, "Inference"),
    ("ESA OC-CCI", 1997, 2021, "Satellite"),
    ("WOA 18", 1950, 2020, "In situ"),
    ("ETOPO 2022", 2022, 2025, "DEM"),
]

# Colors by category
colors = {
    "Satellite": "tab:blue",
    "Reanalysis": "tab:orange",
    "In situ": "tab:green",
    "Sat/In situ": "tab:purple",
    "DEM": "tab:gray",
}

fig, ax = plt.subplots(figsize=(13, 7))

# Draw bars
for i, (name, start, end, category) in enumerate(datasets):

    color = colors[category]

    ax.barh(
        y=i,
        width=end - start,
        left=start,
        height=0.7,
        color=color,
        alpha=0.85,
        edgecolor="black",
        linewidth=0.5
    )

    # Text inside bar
    ax.text(
        x=(start + end) / 2,
        y=i,
        s=f"{name}\n{start}–{end}",
        ha="center",
        va="center",
        color="white",
        fontsize=8,
        fontweight="bold"
    )

# Y axis
ax.set_yticks(range(len(datasets)))
ax.set_yticklabels([""] * len(datasets))  # hide external labels

# X axis years
ax.set_xlim(1970, 2026)

years = np.arange(1970, 2026, 5)

ax.set_xticks(years)
ax.set_xticklabels(years)

# Grid
ax.grid(axis="x", linestyle="--", alpha=0.4)

# Labels
ax.set_xlabel("Year")
ax.set_title("Temporal Coverage of Datasets")

# Invert y-axis so first dataset is on top
ax.invert_yaxis()

# Legend
legend_elements = [
    Patch(facecolor=color, label=category)
    for category, color in colors.items()
]

ax.legend(
    handles=legend_elements,
    loc="lower right"
)

plt.tight_layout()

# Save high-quality figure
# plt.savefig("dataset_timeline.pdf", bbox_inches="tight")

plt.show()
```

```python
import matplotlib.pyplot as plt
from matplotlib.patches import Patch

# ---------------------------------------------------
# DATA
# variable, dataset, start, end, usage
# ---------------------------------------------------

data = [
    ("Salinity",   "GLODAP",         1982, 2021, "Training"),
    ("Salinity",   "ESA SSS CCI",    2010, 2022, "Inference"),
    ("Salinity",   "SODA",           1982, 2010, "Inference"),
    ("Salinity",   "CMEMS MultiObs", 1993, 2025, "Sensitivity"),
    ("Salinity","CMEMS GLORYS",   1993, 2025, "Sensitivity"),

    ("Temperature","GLODAP",         1982, 2021, "Training"),
    #("Temperature","SODA",           1980, 2023, "Sensitivity"),
    ("Temperature","ESA SST CCI",    1982, 2022, "Inference"),
    #("Temperature","OISST",          2022, 2024, "Sensitivity"),

    ("Nutrients (N, P, Si)", "GLODAP",          1982, 2021, "Training"),
    ("Nutrients (N, P, Si)", "WOA 18",          1982, 2022, "Inference"),
    
    ("SSH",   "CMEMS-DUACS SSH",         1993, 2022, "Training"),
    ("SSH",   "CMEMS-DUACS SSH",         1993, 2022, "Inference"),
    
    ("Bottom depth",   "GLODAP",         1982, 2021, "Training"),
    ("Bottom depth",   "ETOPO2022",         1982, 2022, "Inference"),
    
    ("Chl-a",   "GLOBCOLOUR",         1982, 2021, "Training"),
    ("Chl-a",   "GLOBCOLOUR",         1982, 2022, "Inference"),

]

# end: 2024-06-30
#         freq: 1D
#         start: 1993-01-01

# ---------------------------------------------------
# COLORS
# ---------------------------------------------------

colors = {
    "Training": "tab:blue",
    "Inference": "tab:green",
    "Sensitivity": "tab:orange",
}

# ---------------------------------------------------
# CREATE POSITIONS
# ---------------------------------------------------

variables = ["Salinity", "Temperature", "Nutrients (N, P, Si)", "SSH", "Bottom depth", "Chl-a"]

plot_data = []

y = 0
gap = 0.5

for var in variables:

    subset = [d for d in data if d[0] == var]

    for row in subset:
        plot_data.append((y, *row))
        y += 0.7

    y += gap

# ---------------------------------------------------
# PLOT
# ---------------------------------------------------

fig, ax = plt.subplots(figsize=(13, 7))

for y, var, dataset, start, end, usage in plot_data:

    ax.barh(
        y=y,
        width=end - start,
        left=start,
        height=0.5,
        color=colors[usage],
        edgecolor="black",
        linewidth=0.5,
        alpha=0.85
    )

    # label inside bar
    ax.text(
        x=(start + end) / 2,
        y=y,
        s=dataset,
        ha="center",
        va="center",
        color="black",
        fontsize=8,
        fontweight="bold"
    )

# ---------------------------------------------------
# Y AXIS LABELS
# ---------------------------------------------------

group_centers = {}

for var in variables:
    ys = [y for y, v, *_ in plot_data if v == var]
    group_centers[var] = sum(ys) / len(ys)

ax.set_yticks(list(group_centers.values()))
ax.set_yticklabels(list(group_centers.keys()))

# ---------------------------------------------------
# X AXIS
# ---------------------------------------------------

ax.set_xlim(1980, 2025)
ax.set_xticks(range(1980, 2026, 5))
ax.set_xlabel("Year")

# ---------------------------------------------------
# GRID
# ---------------------------------------------------

ax.grid(axis="x", linestyle="--", alpha=0.4)

# ---------------------------------------------------
# LEGEND
# ---------------------------------------------------

legend_elements = [
    Patch(facecolor=colors[key], label=key)
    for key in colors
]

#ax.legend(handles=legend_elements, loc="lower right")

# Remove frame box
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.spines['left'].set_visible(False)

# Keep bottom axis line
ax.spines['bottom'].set_visible(True)
# ---------------------------------------------------
# CLEANUP
# ---------------------------------------------------

ax.invert_yaxis()

plt.tight_layout()
plt.show()
```
